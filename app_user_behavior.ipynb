```python
df['rating_given'] = df['rating_given'].fillna(df['rating_given'].mean())
```

```python
features = [
    'sessions_per_week', 'avg_session_duration_min', 'daily_active_minutes',
    'feature_clicks_per_session', 'notifications_opened_per_week',
    'in_app_search_count', 'pages_viewed_per_session', 'engagement_score',
    'churn_risk_score', 'ads_clicked_last_30_days', 'content_downloads'
]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[features])
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X_scaled)
```

```python
cluster_names = {
    0: "Cluster 0 - High Engagement Users",
    1: "Cluster 1 - Moderate / At-Risk Users",
    2: "Cluster 2 - Low Engagement / Stable Users",
    3: "Cluster 3 - Occasional Users"
}

df['segment_name'] = df['cluster'].map(cluster_names)
```

```python
cluster_means = df.groupby('cluster')[features].mean()
print(cluster_means)
```

```python
pca = PCA(n_components=2)
pca_components = pca.fit_transform(X_scaled)
```

```python
df['PC1'] = pca_components[:, 0]
df['PC2'] = pca_components[:, 1]
```

```python
df.to_csv('clustered_users_with_pca.csv', index=False)
```
